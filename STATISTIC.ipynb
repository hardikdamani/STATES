##### TASK - 4

```python
# P(Food∩Dessert)=Total unique users / Users who ordered both
def probability_both(food_users, dessert_users):
    food_set = set(food_users)
    dessert_set = set(dessert_users)

    both_users = food_set.intersection(dessert_set)
    all_users = food_set.union(dessert_set)

    probability = len(both_users) / len(all_users)

    return probability

food_users = [1, 2, 3, 4, 5, 6]
dessert_users = [3, 4, 5, 7, 8]

probability = probability_both(food_users, dessert_users)

print("Probability of ordering both:", probability)
```

##### TASK - 5

```python
# P(Headphones∣Phone)=
# Users who bought Phone / 
# Users who bought Phone and Headphones

def conditional_probability(phone_users, headphone_users):
    phone_set = set(phone_users)
    headphone_set = set(headphone_users)

    both = phone_set.intersection(headphone_set)

    probability = len(both) / len(phone_set)

    return probability

phone_users = [1, 2, 3, 4, 5, 6, 7, 8]
headphone_users = [2, 3, 5, 7]

probability = conditional_probability(phone_users, headphone_users)

print("Probability of buying headphones given phone:", probability)
```

##### TASK - 2

```python
# 2. Swiggy Premium Restaurant Example

# Given:

# 10% restaurants are premium → P(Premium) = 0.10
# 80% premium restaurants offer discount → P(Discount | Premium) = 0.80
# 20% regular restaurants offer discount → P(Discount | Regular) = 0.20

# First calculate:

# P(Discount)=(0.80×0.10)+(0.20×0.90)
# P(Discount)=0.08+0.18=0.26

# Then:

# P(Premium∣Discount)=
# 0.26
# 0.80×0.10

# =0.307

def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence

prior_premium = 0.10

likelihood = 0.80
evidence = (0.80 * 0.10) + (0.20 * 0.90)

posterior = bayes_posterior(
    prior_premium,
    likelihood,
    evidence
)

print("Probability restaurant is premium:", posterior)
print("Percentage:", posterior * 100)
```
